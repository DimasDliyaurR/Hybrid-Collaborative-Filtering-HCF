```python
def test_similarity_experiment_ti(similarity_experiment, alpha_1 = 1, alpha_2 = 1) :
    similarity = lambda A,B : len( set(A) & set(B) ) / ( len( set(A) & set(B) ) + ( alpha_1*len( set(A) - set(B) ) ) + (alpha_2*len( set(B) - set(A) )) )

    # Uji Coba pertama : Intersection Penuh
    A_1 = [1,2,3,4,5]
    B_1 = A_1.copy()
    s_test_1 = similarity(A_1,B_1)
    se_test_1 = similarity_experiment(A_1,B_1, alpha_1, alpha_2)

    print(f"Uji coba pertama (Intersection Penuh): {'sama' if se_test_1 == s_test_1 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_1}; experiment = {se_test_1}")

    # Uji Coba kedua : Tanpa irisan
    A_2 = [1,2,3,4,5]
    B_2 = [6,7,8,9,10]
    s_test_2 = similarity(A_2,B_2)
    se_test_2 = similarity_experiment(A_2,B_2, alpha_1, alpha_2)
    
    print(f"Uji coba kedua (Tanpa irisan): {'sama' if se_test_2 == s_test_2 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_2}; experiment = {se_test_2}")
    
    # Uji Coba ketiga : panjang berbeda
    A_3 = [1,2,3,4,5]
    B_3 = [1,2,6,7,8,9]
    s_test_3 = similarity(A_3,B_3)
    se_test_3 = similarity_experiment(A_3,B_3, alpha_1, alpha_2)
    
    print(f"Uji coba ketiga (Panjang berbeda): {'sama' if se_test_3 == s_test_3 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_3}; experiment = {se_test_3}")

    # Uji Coba keempat : sebagian overlap
    A_4 = [1,2,3,4]
    B_4 = [3,4,5,6]
    s_test_4 = similarity(A_4,B_4)
    se_test_4 = similarity_experiment(A_4,B_4, alpha_1, alpha_2)

    print(f"Uji coba keempat (Sebagian overlap): {'sama' if se_test_4 == s_test_4 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_4}; experiment = {se_test_4}")

    # Uji Coba kelima : subset
    A_5 = [1,2,3]
    B_5 = [1,2,3,4,5]
    s_test_5 = similarity(A_5,B_5)
    se_test_5 = similarity_experiment(A_5,B_5, alpha_1, alpha_2)

    print(f"Uji coba kelima (Subset): {'sama' if se_test_5 == s_test_5 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_5}; experiment = {se_test_5}")

    # Uji Coba keenam : satu elemen sama
    A_6 = [1]
    B_6 = [1]
    s_test_6 = similarity(A_6,B_6)
    se_test_6 = similarity_experiment(A_6,B_6, alpha_1, alpha_2)

    print(f"Uji coba keenam (Satu elemen sama): {'sama' if se_test_6 == s_test_6 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_6}; experiment = {se_test_6}")

    # Uji Coba ketujuh : satu elemen beda
    A_7 = [1]
    B_7 = [2]
    s_test_7 = similarity(A_7,B_7)
    se_test_7 = similarity_experiment(A_7,B_7, alpha_1, alpha_2)

    print(f"Uji coba ketujuh (Satu elemen beda): {'sama' if se_test_7 == s_test_7 else 'tidak sama'}")
    print(f"Dengan hasil : biasa = {s_test_7}; experiment = {se_test_7}")
```
